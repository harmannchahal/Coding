```python
#WAP to read a line and print the statistics: No. of Uppercase, lowercase letters, digits and alphabets.

line = "Hello123World"

upper = 0
lower = 0
digits = 0
alphabets = 0

for ch in line:
    if ch.isupper():
        upper += 1
        alphabets += 1
    elif ch.islower():
        lower += 1
        alphabets += 1
    elif ch.isdigit():
        digits += 1

print("Line =", line)
print("Uppercase Letters:", upper)
print("Lowercase Letters:", lower)
print("Digits:", digits)
print("Alphabets:", alphabets)
```

```python
#Write a program to print month 'Feb' day 'tuesday' and year 2026 using match case

month = 2
day = 2
year = 2026

match month:
    case 1:
        print("Month: January")
    case 2:
        print("Month: February")
    case 3:
        print("Month: March")
    case _:
        print("Month: Unknown")

match day:
    case 1:
        print("Day: Monday")
    case 2:
        print("Day: Tuesday")
    case 3:
        print("Day: Wednesday")
    case _:
        print("Day: Unknown")

match year:
    case 2026:
        print("Year: 2026")
    case _:
        print("Year: Other")
```

```python
n = 5

for i in range(1, n + 1):     # rows
    for j in range(i):       # columns (stars)
        print("*", end="")
    print()
```

```python
n = 5

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

```python
#inverted right angle traingle
n = 5

for i in range(n, 0, -1):      # rows
    for j in range(i):        # columns (stars)
        print("*", end="")
    print()
```
